### **DICE ROLLING SIMULATOR**

```python
import numpy as np
```

### ***1=minumum,7=maximum values ,20=dice roll 20 times***

```python
dice=np.random.randint(1,7,20)
```

```python
dice
```

```python
print("total rolls:",len(dice))
```

```python
for i in range(1,7):
    print("number", i,"appeared",np.sum(dice==i),"times") #dice==i check krta ha array mn i kitni bar
    #np.sum true value ko count krta ha
```

```python
most=np.argmax(np.bincount(dice))# ya har no ki frequency count krta ha 
print("most frequent",most)
```
